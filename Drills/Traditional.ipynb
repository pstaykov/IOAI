## Classification
Dataset: breast cancer (`sklearn.datasets.load_breast_cancer`), binary classification on 30 numeric features.

```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

data = load_breast_cancer()
df = pd.DataFrame(data.data, columns=data.feature_names)
df['target'] = data.target
```

### EDA
Explore feature distributions and correlations with the target. Which features look most informative?

### Unsupervised Dataprocessing

### Applying models and comparing results
